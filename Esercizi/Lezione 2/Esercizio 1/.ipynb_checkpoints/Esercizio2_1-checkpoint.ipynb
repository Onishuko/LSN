The aim of this exercise is to evaluate numerically the integral

$$
I=\int_0^1 \frac{\pi}{2}\cos\left(\frac{\pi x}{2}\right)dx=1
$$

using two different Monte Carlo techniques: uniform sampling and importance sampling.

#### Uniform sampling

We consider integrals of the form

$$
I=\int g(x)p(x)dx,
$$

where $p(x)$ is a positive and normalized probability density. In the uniform sampling approach, random variables $x_i$ are extracted according to the probability density $p(x)$, and the integral is estimated through the sample mean

$$
G_N=\frac{1}{N}\sum_{i=1}^N g(x_i).
$$

Since $\langle G_N\rangle=I$ for a sufficiently large number of samples, the integral can be expressed as

$$
G_N=I+\epsilon,
$$

where the statistical uncertainty is

$$
\epsilon\simeq\frac{\sigma_I}{\sqrt{N}}.
$$

The variance $\sigma_I^2$ of the estimator is determined by the variance of $g(x)$ with respect to the distribution $p(x)$,

$$
\sigma_I^2= \int g^2(x)p(x)dx-I^2,
$$

The main advantage of this approach is its simplicity and its straightforward extension to multidimensional integrals, making it particularly useful for problems involving high-dimensional domains.

#### Importance sampling

The efficiency of the Monte Carlo estimate can be improved by adding an auxiliary probability density $d(x)$, satisfying

$$
d(x)>0,
\qquad
\int d(x),dx=1.
$$

The integral can then be rewritten as

$$
I=
\int
\frac{g(x)p(x)}{d(x)}
d(x),dx.
$$

The quantity $\frac{g(x)p(x)}{d(x)}$ is therefore averaged by sampling the variables $x_i$ according to $d(x)$. The corresponding estimator is

$$
G_N=
\frac{1}{N}
\sum_{i=1}^N
\frac{g(x_i)p(x_i)}{d(x_i)}.
$$

The variance becomes

$$
\sigma_I^2=
\int
\left[\frac{g(x)p(x)}{d(x)}\right]^2d(x)dx-I^2.
$$

The goal of importance sampling is to choose $d(x)$ so as to minimize the first term of the variance. This can be achieved using the method of Lagrange multipliers. The optimal sampling distribution is found to be proportional to

$$
d(x)\propto |g(x)p(x)|,
$$

In practice, importance sampling can substantially reduce the variance when the sampling distribution is chosen appropriately. It is particularly useful when the integrand has localized regions that contribute significantly to the integral or when the integrand presents singular or rapidly varying behaviour. However, the method can become less effective when $$g(x)p(x)$$ changes sign, since the optimal distribution depends on its absolute value and therefore cannot exploit cancellations between positive and negative contributions.


### Simuation procedure

For the numerical evaluation of the integral, both uniform sampling and importance sampling were implemented. In the uniform sampling case, the variable $x$ is extracted uniformly in the interval $[0,1]$, so that $p(x)=1$. The function used in the Monte Carlo average is therefore simply the integrand,

$$
g(x)=\frac{\pi}{2}\cos\left(\frac{\pi x}{2}\right).
$$

For the importance sampling procedure, a probability density

$$
d(x)=2(1-x)
$$

is introduced. Its cumulative distribution function is

$$
F(x)=\int_0^x 2(1-t),dt=2x-x^2.
$$

Using the inversion method, a uniformly distributed random variable $y\in[0,1]$ is generated and $x$ is obtained by solving $y=F(x)$, giving

$$
x=1-\sqrt{1-y}.
$$

The function entering the Monte Carlo average is consequently the ratio between the original integrand and the importance-sampling distribution,

$$
\frac{g(x)p(x)}{d(x)}=\frac{\frac{\pi}{2}\cos\left(\frac{\pi x}{2}\right)}
{2(1-x)}=\frac{\pi}{4}\frac{\cos\left(\frac{\pi x}{2}\right)}{1-x}.
$$


```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import math
```

```python
ave_unif = np.loadtxt('ave_unif.txt')
ave_cum = np.loadtxt('ave_cum.txt')
err_unif = np.loadtxt('err_unif.txt')
err_cum = np.loadtxt('err_cum.txt')
```

```python
M = 100000
N = 100
L = int(M/N)
x = np.arange(N)
x*=L
```

```python
plt.errorbar(x, ave_unif, yerr=err_unif, ecolor='red')
plt.xlabel('#throws')
plt.ylabel('I')
plt.title('Integral evaluation with uniform sampling')
plt.grid(True)
plt.show()
```

Figure 1. Progressive average of the numerical estimate of the integral using uniform sampling, obtained from $M=10^5$ throws divided into $N=100$ blocks. The associated statistical uncertainty is shown together with the progressive estimate.

```python
plt.errorbar(x, ave_cum, yerr=err_cum, ecolor='red')
plt.xlabel('#throws')
plt.ylabel('I')
plt.title('Integral evaluation with importance sampling')
plt.grid(True)
plt.show()
```

Figure 2. Progressive average of the numerical estimate of the integral using importance sampling, obtained from $M=10^5$ throws divided into $N=100$ blocks. The associated statistical uncertainty is shown together with the progressive estimate.

### Conclusion

The final results obtained with the two sampling techniques are

$$
I_{\mathrm{uniform}}=0.99732000\pm0.00170316,
$$

and

$$
I_{\mathrm{importance}}=1.00011735\pm0.00041232.
$$

We can observe that the importance sampling method provides a statistically more precise result, as shown by the significantly smaller uncertainty. In particular, the uncertainty is reduced by approximately a factor of four compared to uniform sampling. This confirms the effectiveness of importance sampling in reducing the variance of the Monte Carlo estimator when an appropriate sampling distribution is chosen.
