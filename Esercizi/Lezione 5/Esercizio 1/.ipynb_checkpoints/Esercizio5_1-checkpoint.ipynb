The aim of this exercise is to sample, using the Metropolis algorithm, the probability distributions associated with the $1s$ and $2p$ orbitals of the hydrogen atom. The Metropolis algorithm is used to generate configurations distributed according to the corresponding quantum probability densities, while the data blocking method is employed to estimate the average distance of the electron from the origin and its statistical uncertainty.

The numerical results are then compared with the theoretical predictions for the two orbitals,

$$
\langle r\rangle_{1s}=\frac{3}{2}a_0
$$

and

$$
\langle r\rangle_{2p} = 5a_0,
$$

where $a_0$ is the Bohr radius. Finally, the configurations generated by the simulation are used to visualize their spatial distribution in three dimensions. The resulting distributions reproduce the characteristic shapes of the corresponding hydrogen orbitals, providing a visual verification of the probability densities sampled by the Metropolis algorithm.


### The Metropolis Algorithm

The Metropolis algorithm is a sampling method based on rejection techniques and Markov processes, which allows probability density functions to be sampled independently of their analytical complexity and dimensionality. Its main drawback is that consecutive configurations generated by the algorithm are generally correlated. This correlation reduces the amount of independent information contained in the sampled configurations and consequently increases the statistical uncertainty of the measured quantities.

The evolution of the system is described by a function $K(x|y)$, called the **stochastic kernel**, which represents the transition probability density from a state $y$ to a state $x$. If $d_n(x)$ denotes the probability density of the system after $n$ steps, its evolution is governed by the Markov equation

$$
d_{n+1}(x)=
\int K(x|y)d_n(y)\,dy.
$$

A probability density $p(x)$ is said to be **invariant** under the kernel $K$ if it satisfies

$$
p(x)=
\int K(x|y)p(y)\,dy.
$$

A stochastic kernel is said to be **ergodic** if there exists a strictly positive probability density $p(x)$ such that, independently of the initial distribution $d_1(x)$,

$$
p(x)=
\lim_{n\to\infty}d_{n+1}(x).
$$

A kernel is instead said to be **regular** if it preserves the regularity of the probability densities during the evolution, so that an initially well-behaved density generates a sequence of well-behaved densities under repeated application of the kernel. Under suitable conditions, a strictly positive and regular kernel is ergodic if and only if it possesses a strictly positive invariant probability density.

Therefore, if we construct a regular kernel $K$ for which the probability density we want to sample, $p(x)$, is invariant,

$$
p(x)=
\int K(x|y)p(y)\,dy,
$$

then, for any initial distribution $d_1(x)$, the Markov process converges to the desired distribution,

$$
p(x)=
\lim_{n\to\infty}d_{n+1}(x).
$$

The idea of the Metropolis algorithm is therefore to construct a suitable transition probability that satisfies these properties, so that the stationary distribution of the corresponding Markov process coincides with the probability density to be sampled.

To construct such a kernel, the **detailed balance condition** is imposed,

$$
K(x|y)p(y)=
K(y|x)p(x).
$$

This condition guarantees that $p(x)$ is an invariant distribution of the Markov process. The transition kernel is constructed by separating the generation of a trial configuration from its acceptance. In particular, it is written in terms of a **proposal probability** $T(x|y)$ and an **acceptance probability** $A(x|y)$,

$$
K(x|y)=
T(x|y)A(x|y).
$$

The proposal probability $T(x|y)$ describes how a new configuration $x$ is generated starting from the current configuration $y$, while $A(x|y)$ determines whether the proposed move is accepted.

Given a proposal probability $T(x|y)$, a possible choice for the acceptance probability is

$$
A(x|y)=
\min\left[1,q(x|y)\right],
$$

where

$$
q(x|y)=
\frac{T(y|x)p(x)}
{T(x|y)p(y)}
\geq 0.
$$

With this choice, the resulting transition kernel satisfies the detailed balance condition and therefore admits $p(x)$ as an invariant distribution.

In the particular case of a symmetric proposal probability,

$$
T(x|y)=T(y|x),
$$

the acceptance probability simplifies to

$$
A(x|y)=
\min\left[1,\frac{p(x)}{p(y)}\right].
$$

Thus, starting from an initial configuration, the Metropolis algorithm proposes a new configuration according to $T(x|y)$ and accepts it with probability $A(x|y)$. Repeating this procedure generates a Markov chain whose configurations, after a sufficiently long equilibration period, are distributed according to the desired probability density $p(x)$.


#### Hydrogen atom
The wave functions of the eigenstates of an electron in a Hydrogen atom (this is the only atom for which the Schroedinger equation has been solved exactly) are expressed in terms of spherical harmonics and generalized Laguerre polynomials. It is convenient to use spherical coordinates, and the wave function can be separated into functions of each coordinate:

$$
\Psi_{n,l,m}(r,\theta,\phi)=
\sqrt{\left(\frac{2}{na_0}\right)^3 \frac{(n-l-1)!}{2n[(n+l)!]}}
e^{-r/na_0}\left(\frac{2r}{na_0}\right)^l L_{n-l-1}^{2l+1}\left(\frac{2r}{na_0}\right)
Y_l^m(\theta,\phi)
$$

where $a_0=4\pi\epsilon_0\hbar^2/m_e e^2=0.0529$ nm is the Bohr radius, $L_{n-l-1}^{2l+1}$ are the generalized Laguerre polynomials of degree $n-l-1$, $n=1,2,...$ is the principal quantum number, $l=0,1, ..., n-1$ the azimuthal quantum number, $m=-l, -l+1, ..., l-1, l$ the magnetic quantum number.

For example, the ground state wave function is:
$$
\Psi_{1,0,0}(r,\theta,\phi)=
\frac{a_0^{-3/2}}{\sqrt{\pi}} e^{-r/a_0} 
$$
whereas one of the three $2p$ excited state is:
$$
\Psi_{2,1,0}(r,\theta,\phi)=
\frac{a_0^{-5/2}}{8}\sqrt{\frac{2}{\pi}} r e^{-r/2a_0} \cos(\theta)
$$
Expectation values for the radius turns out to be exactly:
$$
\left\langle r \right\rangle_{\Psi_{1,0,0}} = \frac{3}{2}a_0
\quad
\left\langle r \right\rangle_{\Psi_{2,1,0}} = 5 a_0
$$

### Simulation Setup

For each hydrogen orbital, two different proposal distributions were considered for the Metropolis algorithm. The first one is a uniform distribution over an interval of width $2\delta$,

$$
T(x|y)\sim\mathcal{U}(-\delta,\delta),
$$

while the second one is a Gaussian distribution centered on the current position,

$$
T(x|y)\sim\mathcal{N}(0,\delta).
$$

The parameter $\delta$, which determines the proposal width, was tuned separately for each orbital and proposal distribution in order to obtain an acceptance rate of approximately $50\%$. The values obtained through this fine-tuning procedure are reported below:

| Proposal distribution |         $1s$  |       $2p$   |
| --------------------- | ------------: | -----------: |
| Uniform               | $\delta=1.20$ | $\delta=3.00$ |
| Gaussian              | $\delta=0.75$ | $\delta=1.85$ |

For both proposal distributions, the initial position was chosen differently for the two orbitals. For the $1s$ orbital, the chain was initialized at

$$
\mathbf r_0=(0,0,0),
$$

whereas for the $2p$ orbital the initial position was chosen as

$$
\mathbf r_0=(0,0,1).
$$

The latter choice avoids starting from the origin, where the probability density of the $2p$ orbital vanishes.

For each case, the simulation was divided into two successive stages. First, a Metropolis chain of $10^4$ steps was generated as an **equilibration phase**, allowing the system to move away from the arbitrary initial configuration and approach the equilibrium distribution. The final configuration reached during this phase was then used as the starting point for a second Metropolis simulation consisting of $10^6$ steps.

The positions generated by the second Metropolis chain were subsequently converted into their distances from the origin,

$$
r=\sqrt{x^2+y^2+z^2},
$$

which were used to estimate the mean radial distance. Finally, the resulting data were analysed using the data blocking method, with a total of $M=10^6$ sampled configurations divided into $N=100$ blocks. This procedure provides both the estimate of the mean distance from the origin and its statistical uncertainty.


```python
def plot(mean_1s_file, error_1s_file,
                  mean_2p_file, error_2p_file):

    # Lettura dei dati
    mean_1s = np.loadtxt(mean_1s_file)
    error_1s = np.loadtxt(error_1s_file)

    mean_2p = np.loadtxt(mean_2p_file)
    error_2p = np.loadtxt(error_2p_file)

    # Numero di blocchi
    N_1s = len(mean_1s)
    N_2p = len(mean_2p)

    blocks_1s = np.arange(1, N_1s + 1)
    blocks_2p = np.arange(1, N_2p + 1)

    # Valori teorici
    theoretical_1s = 1.5
    theoretical_2p = 5.0

    # Figura 1x2
    fig, ax = plt.subplots(1, 2, figsize=(14, 5))

    # 1s
    ax[0].errorbar(
        blocks_1s,
        mean_1s,
        yerr=error_1s,
        ecolor = "red",
        markersize=3,
        capsize=2,
        alpha=0.7,
        label='Metropolis'
    )

    ax[0].axhline(
        theoretical_1s,
        linestyle='--',
        label=r'Theoretical $\langle r\rangle = 3/2a_0$'
    )

    ax[0].set_title(r'$1s$ state')
    ax[0].set_xlabel('Number of blocks')
    ax[0].set_ylabel(r'$\langle r\rangle$')
    ax[0].grid(True)
    ax[0].legend()

    # 2p
    ax[1].errorbar(
        blocks_2p,
        mean_2p,
        yerr=error_2p,
        ecolor = "red",
        markersize=3,
        capsize=2,
        alpha=0.7,
        label='Metropolis'
    )

    ax[1].axhline(
        theoretical_2p,
        linestyle='--',
        label=r'Theoretical $\langle r\rangle = 5a_0$'
    )

    ax[1].set_title(r'$2p$ state')
    ax[1].set_xlabel('Number of blocks')
    ax[1].set_ylabel(r'$\langle r\rangle$')
    ax[1].grid(True)
    ax[1].legend()

    plt.tight_layout()
    plt.show()

    
```

```python
plot("mean_1s_uniform.dat", "error_1s_uniform.dat", "mean_2p_uniform.dat", "error_2p_uniform.dat")
```

Figure 1. Progressive estimates of the average distance from the origin, $\langle r\rangle$, for the $1s$ orbital (left) and the $2p$ orbital (right). The configurations are generated using a uniform proposal distribution, with statistically correlated successive measurements.


```python
plot("mean_1s_gaussian.dat", "error_1s_gaussian.dat", "mean_2p_gaussian.dat", "error_2p_gaussian.dat")
```

Figure 2. Progressive estimates of the average distance from the origin, $\langle r\rangle$, for the $1s$ orbital (left) and the $2p$ orbital (right). The configurations are generated using a Gaussian proposal distribution, with statistically correlated successive measurements.


```python
def plot_distribution(file_1s, file_2p):

    x_1s, y_1s, z_1s = np.loadtxt(file_1s, usecols=(0, 1, 2), unpack=True, skiprows=1)
    x_2p, y_2p, z_2p = np.loadtxt(file_2p, usecols=(0, 1, 2), unpack=True, skiprows=1)

    fig = plt.figure(figsize=(12,6))
    
    ax1 = fig.add_subplot(1, 2, 1, projection = '3d')
    ax1.scatter(x_1s, y_1s, z_1s, marker='.', alpha=0.5)
    ax1.set_title(r'Sampling $1s$')
    ax1.set_xlabel('x')
    ax1.set_ylabel('y')
    ax1.set_zlabel('z')

    ax2 = fig.add_subplot(1, 2, 2, projection = '3d')
    ax2.scatter(x_2p, y_2p, z_2p, marker='.', alpha=0.5)
    ax2.view_init(elev=0, azim=90)
    ax2.set_title(r'Sampling $2p$')
    ax2.set_xlabel('x')
    ax2.set_ylabel('y')
    ax2.set_zlabel('z')

    plt.tight_layout()
    plt.show()
    
```

```python
plot_distribution("sampling_1s_uniform.dat", "sampling_2p_uniform.dat")
```

Figure 3. Three-dimensional distribution of the sampled positions for the $1s$ orbital (left) and the $2p$ orbital (right). The spatial distributions accurately reproduce the characteristic shapes of the corresponding hydrogen orbitals.


```python
plot_distribution("sampling_1s_gaussian.dat", "sampling_2p_gaussian.dat")
```

Figure 4. Three-dimensional distribution of the sampled positions for the $1s$ orbital (left) and the $2p$ orbital (right). The spatial distributions accurately reproduce the characteristic shapes of the corresponding hydrogen orbitals.


### Conclusions

The results show that the Metropolis algorithm successfully samples the probability densities of both the $1s$ and $2p$ hydrogen orbitals, with the estimated average distances being consistent with the corresponding theoretical predictions. The three-dimensional distributions of the sampled configurations also accurately reproduce the characteristic spatial shapes of the two orbitals, confirming the effectiveness of the method.
